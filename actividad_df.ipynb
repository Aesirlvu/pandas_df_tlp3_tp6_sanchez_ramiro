```python
import pandas as pd
import sqlite3

#Aqui agrega la conexión a la base de datos

conn = sqlite3.connect('Salaries.sqlite')

#Crea el dataframe con los datos del datasets

df = pd.read_sql_query('SELECT * FROM Salaries', conn)

conn.close()
```

1. **Top 10 empleados con mayor salario total (incluyendo beneficios)**
Extrae e imprime un top 10 de expleados con mayor salario incluyendo los beneficios 


```python
#todo
# ordenamos en base a la columna 'TotalPayBenefits' de forma descendente y tomamos los primeros 10 registros
top_10_empleados = df.sort_values(by='TotalPayBenefits', ascending=False).head(10)

print(top_10_empleados[['EmployeeName','TotalPayBenefits']])
```

2. **Filtrar empleados con más de 50,000 en horas extra, la columna a tratar es: (OvertimePay)**

```python
# para aplicar un filtro necesitamos convertir la columna 'OvertimePay' a un tipo numérico
# de lo contrario obtenemos un ValueError.
df['OvertimePay'] = pd.to_numeric(df['OvertimePay'], errors='coerce')
```

```python
#todo
empleados_horas_extra = df[df['OvertimePay'] > 50000]

# mostramos los primeros 10 registros luego de aplicar el filtro
print(empleados_horas_extra[['EmployeeName','OvertimePay']].head(10))
print(f'\nempleados con más de 50k en horas extra: {len(empleados_horas_extra)}')
```

3. **Contar cuántos empleados únicos hay por año**

```python
#todo
empleados_por_año = df.groupby('Year')['EmployeeName'].nunique()
print(f'cantidad de empleados por año: \n{empleados_por_año}')
```

4. **Ver cuántos cargos únicos (JobTitle) existen y los 5 más comunes**

```python
cargos_unicos = df['JobTitle'].nunique()
print(f'existen: {cargos_unicos} cargos únicos')
```

```python
#todo
# con size() contamos la cantidad de registros por cada grupo
# y sort_values para ordenarlos de forma descendente,
# para conseguir los 5 cargos más comunes
cantidad_cargos_unicos = df.groupby('JobTitle').size().sort_values(ascending=False).head(5)

print(f'cargos más comunes: \n{cantidad_cargos_unicos}')
```

5. **Mostrar el salario total (TotalPay) promedio por año**

```python
#todo
# agrupamos por año y la columna 'TotalPay', luego aplicamos la función mean() para el promedio,
# y round(2) para redondear a 2 decimales
salario_tota_promedio_por_año = df.groupby('Year')['TotalPay'].mean().round(2)

print(salario_tota_promedio_por_año)
```

6. **Exporta todos los datos a un CSV y muestra los ultimos 10 valores**

```python
#todo
# para exportar el dataframe a un archivo CSV, usamos el método to_csv()
# index en false para no incluir el índice en el CSV
df.to_csv('Salaries.csv', index=False)

# y mostramos los ultimos 10 registros
df.tail(10)
```
